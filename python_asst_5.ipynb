```python
numbers = [10, 15, 20, 25, 30, 35]

even = 0
odd = 0

for i in numbers:
    if i % 2 == 0:
        even = even + 1
    else:
        odd = odd + 1

print("Even numbers =", even)
print("Odd numbers =", odd)
```

```python
numbers = [10, -5, 0, 20, -10, 0, 15]

positive = 0
negative = 0
zero = 0

for i in numbers:
    if i > 0:
        positive = positive + 1
    elif i < 0:
        negative = negative + 1
    else:
        zero = zero + 1

print("Positive =", positive)
print("Negative =", negative)
print("Zero =", zero)
```

```python
num = int(input("Enter a number: "))

reverse = 0

while num != 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

print("Reverse =", reverse)
```

```python
num = int(input("Enter a number: "))

original = num
reverse = 0

while num != 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

if original == reverse:
    print("Palindrome")
else:
    print("Not a Palindrome")
```
